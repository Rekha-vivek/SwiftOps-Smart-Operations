```python
#manual method to find outliers
Q1=so_df['Current_Stock'].quantile(0.25)
Q3=so_df['Current_Stock'].quantile(0.75)

IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

cs_outliers=so_df[
    (so_df['Current_Stock']< lower_bound)|
    (so_df['Current_Stock']>upper_bound)
]
cs_outliers
```

```python
#easy method to find outliers using define function

def find_outliers(df,column):
    Q1=df[column].quantile(0.25)
    Q3=df[column].quantile(0.75)

    IQR=Q3-Q1 

    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR

    return df[
        (df[column]<lower_bound)|
        (df[column]>upper_bound)]
```
